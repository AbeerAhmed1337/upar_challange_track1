```python
# ===== 4. LABELS + IMAGE PATHS (numpy/pandas only) =====
import os, numpy as np, pandas as pd

ATTR_NAMES = ["Age-Young","Age-Adult","Age-Old","Gender-Female","Hair-Length-Short","Hair-Length-Long","Hair-Length-Bald",
 "UpperBody-Length-Short","UpperBody-Color-Black","UpperBody-Color-Blue","UpperBody-Color-Brown","UpperBody-Color-Green",
 "UpperBody-Color-Grey","UpperBody-Color-Orange","UpperBody-Color-Pink","UpperBody-Color-Purple","UpperBody-Color-Red",
 "UpperBody-Color-White","UpperBody-Color-Yellow","UpperBody-Color-Other","LowerBody-Length-Short","LowerBody-Color-Black",
 "LowerBody-Color-Blue","LowerBody-Color-Brown","LowerBody-Color-Green","LowerBody-Color-Grey","LowerBody-Color-Orange",
 "LowerBody-Color-Pink","LowerBody-Color-Purple","LowerBody-Color-Red","LowerBody-Color-White","LowerBody-Color-Yellow",
 "LowerBody-Color-Other","LowerBody-Type-Trousers&Shorts","LowerBody-Type-Skirt&Dress","Accessory-Backpack","Accessory-Bag",
 "Accessory-Glasses-Normal","Accessory-Glasses-Sun","Accessory-Hat"]
assert len(ATTR_NAMES) == 40

DATA = f'{REPO}/data'
ANN = f'{DATA}/annotations/task1'
# csv paths may be relative to data/, to the repo, or to a dataset folder: try each root
ROOTS = [DATA, REPO, f'{DATA}/Market1501', f'{DATA}/PA100k', f'{DATA}/PA100k/release_data/release_data',
         f'{DATA}/PETA', f'{DATA}/PETA/images', '']
_root_cache = {}

def resolve(p):
    p = str(p).replace('\\', '/')
    key = p.split('/')[0]
    r = _root_cache.get(key)
    if r is not None:
        return os.path.join(r, p)
    for r in ROOTS:
        if os.path.exists(os.path.join(r, p)):
            _root_cache[key] = r
            return os.path.join(r, p)
    return None

def ds_name(s):
    s = s.lower()
    if 'market' in s: return 'Market1501'
    if 'pa100' in s: return 'PA100k'
    if 'peta' in s: return 'PETA'
    return 'other'

def load_split(split):
    df = pd.read_csv(f'{ANN}/{split}/gt.csv')
    df = df.rename(columns={df.columns[0]: 'image'})
    if set(ATTR_NAMES) <= set(df.columns):
        df = df[['image'] + ATTR_NAMES]                     # force the official attribute order
    else:
        print('WARNING: column names differ from the official list, assuming the 40 columns are in official order')
        df = df.iloc[:, :41]
    full = [resolve(p) for p in df['image']]
    found = np.array([f is not None and os.path.exists(f) for f in full])
    names = np.array([ds_name(str(p) + ' ' + (f or '')) for p, f in zip(df['image'], full)])
    print(f'[{split}] rows={len(df)} found on disk={int(found.sum())}')
    print(pd.crosstab(names, found, rownames=['dataset'], colnames=['found']).to_string())
    if found.sum() == 0:
        raise RuntimeError(f'No image resolved. First csv paths: {list(df["image"][:3])}; roots tried: {ROOTS}')
    raw = df[ATTR_NAMES].to_numpy(dtype=np.float32)
    raw = np.nan_to_num(raw, nan=-1.0)
    valid = ((raw == 0) | (raw == 1)).astype(np.float32)    # anything else (e.g. -1, 2) = not annotated -> ignored
    lab = (raw == 1).astype(np.float32)
    out = dict(paths=np.array(full, dtype=object)[found], lab=lab[found], valid=valid[found], ds=names[found])
    print(f'[{split}] fraction of labels marked invalid: {1 - out["valid"].mean():.3f}')
    return out

train_raw = load_split('train')
val_raw = load_split('val')
```

```python
def par_scores(probs, lab, valid, th):
    """mA, instance-F1, Challenge_Avg (harmonic mean) for ONE dataset. th: scalar or (40,). Invalid labels are ignored."""
    pred = probs > th
    lab = lab > 0.5
    v = valid > 0.5
    tp = (pred & lab & v).sum(0); fn = (~pred & lab & v).sum(0)
    tn = (~pred & ~lab & v).sum(0); fp = (pred & ~lab & v).sum(0)
    keep = (tp + fn + tn + fp) > 0
    rec = tp / np.maximum(tp + fn, 1); spec = tn / np.maximum(tn + fp, 1)
    mA = float(((rec + spec) / 2)[keep].mean())
    P, G = pred & v, lab & v
    inter = (P & G).sum(1).astype(np.float64)
    prec = inter / np.maximum(P.sum(1), 1e-8)
    rec_i = inter / np.maximum(G.sum(1), 1e-8)
    f1 = float(np.mean(2 * prec * rec_i / np.maximum(prec + rec_i, 1e-8)))
    return mA, f1, 2 * mA * f1 / max(mA + f1, 1e-8)

def mean_challenge(probs, lab, valid, masks, th):
    """The leaderboard averages per-dataset scores, so we do too."""
    return float(np.mean([par_scores(probs[m], lab[m], valid[m], th)[2] for m in masks.values()]))

def tune_thresholds(probs, lab, valid, masks, grid=np.arange(0.20, 0.801, 0.05), passes=2):
    """Coordinate ascent on mean-over-datasets Challenge_Avg. Search range is narrow and ties go to 0.5 to limit overfitting."""
    n = probs.shape[1]
    th = np.full(n, 0.5)
    obj = lambda t: mean_challenge(probs, lab, valid, masks, t) - 1e-3 * np.abs(t - 0.5).mean()
    best = obj(th)
    for _ in range(passes):
        for j in range(n):
            keep_g = th[j]
            for g in grid:
                th[j] = g
                s = obj(th)
                if s > best + 1e-9:
                    best, keep_g = s, g
            th[j] = keep_g
    return th, mean_challenge(probs, lab, valid, masks, th)
```

```python
# ===== 6. CONFIG, SPLITS, AUGMENTATION, LOADERS =====
import io, math, random, time, copy, glob, json
import torch, torch.nn as nn, torch.nn.functional as F
from PIL import Image
import torchvision.transforms as T
from torch.utils.data import Dataset, DataLoader

# ---- knobs ----
QUICK = False            # True = ~10 min smoke test of the WHOLE pipeline (train -> zip -> clean-venv test). Do this first.
MODEL_NAME = 'convnext_base.fb_in22k_ft_in1k'     # faster: 'convnext_small.fb_in22k_ft_in1k'
IMG_H, IMG_W = 256, 128
BATCH, EPOCHS = 32, 3
LR_BB, LR_HEAD, WD = 8e-5, 8e-4, 0.05
EMA_DECAY, LABEL_SMOOTH = 0.999, 0.05
HOLDOUT_PER_DS = 2000    # val images per dataset kept for thresholds/model selection; the rest of val is used for training
SEED = 0
if QUICK:
    EPOCHS, HOLDOUT_PER_DS = 1, 150
CKPT_DIR = f'{CKPT_ROOT}/{"quick" if QUICK else "full"}'
os.makedirs(CKPT_DIR, exist_ok=True)

random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
torch.backends.cudnn.benchmark = True
device = torch.device('cuda')

# ---- split: hold out a per-dataset slice of val, train on train + rest of val (allowed by the rules) ----
rng = np.random.RandomState(SEED)
hold = np.zeros(len(val_raw['ds']), dtype=bool)
for d in np.unique(val_raw['ds']):
    idx = np.where(val_raw['ds'] == d)[0]
    n = min(HOLDOUT_PER_DS, max(1, len(idx) // 3))
    hold[rng.choice(idx, n, replace=False)] = True

def cat(a, b): return np.concatenate([a, b])
tr = {k: cat(train_raw[k], val_raw[k][~hold]) for k in train_raw}
ho = {k: val_raw[k][hold] for k in val_raw}
if QUICK:
    sel = rng.choice(len(tr['ds']), 2400, replace=False)
    tr = {k: v[sel] for k, v in tr.items()}
print('train images:', len(tr['ds']), '| holdout images:', len(ho['ds']))
print('train per dataset:', dict(zip(*np.unique(tr['ds'], return_counts=True))))
pos = (tr['lab'] * tr['valid']).sum(0); neg = ((1 - tr['lab']) * tr['valid']).sum(0)
rare = [(ATTR_NAMES[j], int(pos[j])) for j in range(40) if pos[j] < 100]
print('attributes with < 100 positive training labels (weak or hopeless on test):', rare)
pos_w = torch.tensor(np.clip(np.sqrt(neg / np.maximum(pos, 1)), 1, 8), dtype=torch.float32, device=device)
bias_init = torch.tensor(np.log((pos + 1) / (neg + 1)), dtype=torch.float32)

# ---- augmentation (no hue / grayscale: colour is a label) ----
class RandomLowRes:
    def __init__(self, p=0.3, lo=0.35): self.p, self.lo = p, lo
    def __call__(self, img):
        if random.random() > self.p: return img
        w, h = img.size; s = random.uniform(self.lo, 1.0)
        return img.resize((max(8, int(w * s)), max(16, int(h * s))), Image.BILINEAR).resize((w, h), Image.BILINEAR)

class RandomJPEG:
    def __init__(self, p=0.2, q=(30, 90)): self.p, self.q = p, q
    def __call__(self, img):
        if random.random() > self.p: return img
        buf = io.BytesIO(); img.save(buf, 'JPEG', quality=random.randint(*self.q)); buf.seek(0)
        return Image.open(buf).convert('RGB')

norm = T.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
train_tf = T.Compose([
    RandomLowRes(0.3), RandomJPEG(0.2),
    T.Resize((IMG_H, IMG_W)),
    T.RandomApply([T.GaussianBlur(5, (0.1, 1.5))], p=0.15),
    T.RandomAffine(degrees=5, translate=(0.05, 0.05), scale=(0.9, 1.1)),
    T.RandomHorizontalFlip(),
    T.ColorJitter(0.3, 0.3, 0.3, 0.0),
    T.ToTensor(), norm,
    T.RandomErasing(p=0.2, scale=(0.02, 0.12)),
])
eval_tf = T.Compose([T.Resize((IMG_H, IMG_W)), T.ToTensor(), norm])

class PARDataset(Dataset):
    def __init__(self, d, tf): self.d, self.tf = d, tf
    def __len__(self): return len(self.d['paths'])
    def __getitem__(self, i):
        try: img = Image.open(self.d['paths'][i]).convert('RGB')
        except Exception: img = Image.new('RGB', (IMG_W, IMG_H))
        return self.tf(img), torch.from_numpy(self.d['lab'][i]), torch.from_numpy(self.d['valid'][i])

NW = min(8, os.cpu_count() or 2)
train_loader = DataLoader(PARDataset(tr, train_tf), batch_size=BATCH, shuffle=True, drop_last=True, num_workers=NW,
                          pin_memory=True, persistent_workers=True, prefetch_factor=4)
hold_loader = DataLoader(PARDataset(ho, eval_tf), batch_size=64, shuffle=False, num_workers=NW, pin_memory=True)
hold_masks = {d: ho['ds'] == d for d in np.unique(ho['ds'])}
print('workers:', NW, '| steps/epoch:', len(train_loader))
```

```python
# ===== 7. MODEL, LOSS, EMA, EVALUATION =====
import timm

class PARNet(nn.Module):
    def __init__(self, name, n=40, pretrained=True, drop_path=0.1, drop=0.2):
        super().__init__()
        self.backbone = timm.create_model(name, pretrained=pretrained, num_classes=0, drop_path_rate=drop_path)
        self.drop = nn.Dropout(drop)
        self.fc = nn.Linear(self.backbone.num_features, n)
    def forward(self, x):
        return self.fc(self.drop(self.backbone(x)))

def masked_bce(logits, y, valid, pos_weight, eps):
    y_s = y * (1 - eps) + 0.5 * eps
    loss = F.binary_cross_entropy_with_logits(logits, y_s, pos_weight=pos_weight, reduction='none') * valid
    return loss.sum() / valid.sum().clamp(min=1)

class EMA:
    def __init__(self, model, decay):
        self.module = copy.deepcopy(model).eval(); self.decay = decay
        for p in self.module.parameters(): p.requires_grad_(False)
    @torch.no_grad()
    def update(self, model, step):
        d = min(self.decay, (1 + step) / (10 + step))
        e = list(self.module.parameters()); m = [p.detach() for p in model.parameters()]
        torch._foreach_mul_(e, d); torch._foreach_add_(e, m, alpha=1 - d)

def lr_scale(step, total, warmup=300):
    if step < warmup: return (step + 1) / warmup
    p = (step - warmup) / max(1, total - warmup)
    return 0.01 + 0.99 * 0.5 * (1 + math.cos(math.pi * p))

def make_optimizer(model):
    g = {k: [] for k in ['bb_d', 'bb_n', 'hd_d', 'hd_n']}
    for n, p in model.named_parameters():
        nd = p.ndim <= 1 or n.endswith('.bias') or 'gamma' in n
        g[('hd_' if n.startswith('fc.') else 'bb_') + ('n' if nd else 'd')].append(p)
    groups = [dict(params=g['bb_d'], weight_decay=WD, base_lr=LR_BB, lr=LR_BB),
              dict(params=g['bb_n'], weight_decay=0.0, base_lr=LR_BB, lr=LR_BB),
              dict(params=g['hd_d'], weight_decay=WD, base_lr=LR_HEAD, lr=LR_HEAD),
              dict(params=g['hd_n'], weight_decay=0.0, base_lr=LR_HEAD, lr=LR_HEAD)]
    return torch.optim.AdamW([x for x in groups if len(x['params'])])

@torch.no_grad()
def predict_logits(m, loader, flip=True, amp=True):
    m.eval(); outs = []
    for x, *_ in loader:
        x = x.to(device, non_blocking=True).contiguous(memory_format=torch.channels_last)
        with torch.autocast('cuda', dtype=torch.float16, enabled=amp):
            lg = m(x).float()
            if flip: lg = (lg + m(torch.flip(x, dims=[3])).float()) / 2
        outs.append(lg.cpu())
    return torch.cat(outs).numpy()

def evaluate(m):
    probs = 1 / (1 + np.exp(-predict_logits(m, hold_loader)))
    th0 = np.full(40, 0.5)
    at05 = {d: par_scores(probs[k], ho['lab'][k], ho['valid'][k], th0) for d, k in hold_masks.items()}
    th, tuned = tune_thresholds(probs, ho['lab'], ho['valid'], hold_masks)
    att = {d: par_scores(probs[k], ho['lab'][k], ho['valid'][k], th) for d, k in hold_masks.items()}
    return dict(avg05=np.mean([r[2] for r in at05.values()]), tuned=tuned, per_ds=att, th=th)
```

```python
# ===== 8. TRAIN (resumable: re-run this cell after a disconnect, it continues from the last finished epoch) =====
model = PARNet(MODEL_NAME).to(device).to(memory_format=torch.channels_last)
with torch.no_grad(): model.fc.bias.copy_(bias_init.to(device))
ema = EMA(model, EMA_DECAY)
opt = make_optimizer(model)
scaler = torch.amp.GradScaler('cuda')
steps_per_epoch = len(train_loader); total_steps = steps_per_epoch * EPOCHS
start_epoch, best_score = 0, -1.0

last_path = f'{CKPT_DIR}/last.pt'
if os.path.exists(last_path):
    ck = torch.load(last_path, map_location='cpu', weights_only=False)
    model.load_state_dict(ck['model']); ema.module.load_state_dict(ck['ema'])
    opt.load_state_dict(ck['opt']); scaler.load_state_dict(ck['scaler'])
    start_epoch, best_score = ck['epoch'] + 1, ck['best']
    print(f'resumed: starting at epoch {start_epoch + 1}/{EPOCHS}')

for epoch in range(start_epoch, EPOCHS):
    model.train(); t0 = time.time(); run = 0.0
    for i, (x, y, v) in enumerate(train_loader):
        gstep = epoch * steps_per_epoch + i
        s = lr_scale(gstep, total_steps)
        for g in opt.param_groups: g['lr'] = g['base_lr'] * s
        x = x.to(device, non_blocking=True).contiguous(memory_format=torch.channels_last)
        y, v = y.to(device, non_blocking=True), v.to(device, non_blocking=True)
        with torch.autocast('cuda', dtype=torch.float16):
            logits = model(x)
        loss = masked_bce(logits.float(), y, v, pos_w, LABEL_SMOOTH)
        opt.zero_grad(set_to_none=True)
        scaler.scale(loss).backward()
        scaler.unscale_(opt); nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        scaler.step(opt); scaler.update()
        ema.update(model, gstep)
        run += loss.item() if i % 20 == 0 else 0
        if i % 200 == 0 and i > 0:
            el = time.time() - t0
            print(f'ep {epoch+1} step {i}/{steps_per_epoch} loss~{run / (i // 20 + 1):.4f} '
                  f'{i * BATCH / el:.0f} img/s | epoch ETA {(steps_per_epoch - i) * el / i / 60:.1f} min', flush=True)
    r = evaluate(ema.module)
    print(f'=== epoch {epoch+1}: holdout Challenge_Avg @0.5 = {r["avg05"]*100:.2f} | with tuned thresholds = {r["tuned"]*100:.2f}')
    for d, (mA, f1, av) in r['per_ds'].items():
        print(f'    {d:11s} mA={mA*100:.2f}  instF1={f1*100:.2f}  avg={av*100:.2f}')
    if r['tuned'] > best_score:
        best_score = r['tuned']
        torch.save({'ema': {k: v.detach().half().cpu() for k, v in ema.module.state_dict().items()},
                    'th': torch.tensor(r['th']), 'score': best_score, 'epoch': epoch}, f'{CKPT_DIR}/best.pt')
        print('    saved best.pt')
    torch.save({'model': model.state_dict(), 'ema': ema.module.state_dict(), 'opt': opt.state_dict(),
                'scaler': scaler.state_dict(), 'epoch': epoch, 'best': best_score}, last_path)
print('done. best holdout score:', best_score)
```

```python
%%writefile /content/submission/run.py
import os, sys, json
HERE = os.path.dirname(os.path.abspath(__file__))
sys.path.insert(0, os.path.join(HERE, 'libs'))        # vendored timm (not part of the runtime image)

import numpy as np
import torch, torch.nn as nn
from PIL import Image
from concurrent.futures import ThreadPoolExecutor
import timm

BATCH_SIZE = 64
ATTR_NAMES = ["Age-Young","Age-Adult","Age-Old","Gender-Female","Hair-Length-Short","Hair-Length-Long","Hair-Length-Bald",
 "UpperBody-Length-Short","UpperBody-Color-Black","UpperBody-Color-Blue","UpperBody-Color-Brown","UpperBody-Color-Green",
 "UpperBody-Color-Grey","UpperBody-Color-Orange","UpperBody-Color-Pink","UpperBody-Color-Purple","UpperBody-Color-Red",
 "UpperBody-Color-White","UpperBody-Color-Yellow","UpperBody-Color-Other","LowerBody-Length-Short","LowerBody-Color-Black",
 "LowerBody-Color-Blue","LowerBody-Color-Brown","LowerBody-Color-Green","LowerBody-Color-Grey","LowerBody-Color-Orange",
 "LowerBody-Color-Pink","LowerBody-Color-Purple","LowerBody-Color-Red","LowerBody-Color-White","LowerBody-Color-Yellow",
 "LowerBody-Color-Other","LowerBody-Type-Trousers&Shorts","LowerBody-Type-Skirt&Dress","Accessory-Backpack","Accessory-Bag",
 "Accessory-Glasses-Normal","Accessory-Glasses-Sun","Accessory-Hat"]

_CFG = json.load(open(os.path.join(HERE, 'config.json')))
H, W = _CFG['h'], _CFG['w']
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')


class PARNet(nn.Module):                       # identical structure to training, so the state dict keys match
    def __init__(self, name, n=40):
        super().__init__()
        self.backbone = timm.create_model(name, pretrained=False, num_classes=0)
        self.drop = nn.Dropout(0.0)
        self.fc = nn.Linear(self.backbone.num_features, n)

    def forward(self, x):
        return self.fc(self.drop(self.backbone(x)))


_MODEL = None
_SHIFT = None
_IDX = None
_MEAN = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1)
_STD = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1)
_POOL = ThreadPoolExecutor(8)


def load_model():
    global _MODEL, _SHIFT, _MEAN, _STD
    m = PARNet(_CFG['model'])
    m.load_state_dict(torch.load(os.path.join(HERE, 'weights.pt'), map_location='cpu', weights_only=True))
    _MODEL = m.to(DEVICE).eval().to(memory_format=torch.channels_last)
    th = np.load(os.path.join(HERE, 'thresholds.npy')).astype(np.float32).clip(0.05, 0.95)
    # The platform thresholds at 0.5. Shifting the logit by logit(th) makes "p > 0.5" mean "raw prob > tuned threshold".
    _SHIFT = torch.logit(torch.from_numpy(th)).to(DEVICE)
    _MEAN, _STD = _MEAN.to(DEVICE), _STD.to(DEVICE)


def _load(path):
    try:
        img = Image.open(path).convert('RGB').resize((W, H), Image.BILINEAR)   # same as T.Resize in training
    except Exception:
        img = Image.new('RGB', (W, H))
    return np.asarray(img, dtype=np.float32) / 255.0


@torch.no_grad()
def predict_batch(samples):
    global _IDX
    if _MODEL is None:
        load_model()
    if _IDX is None:
        names = samples[0].get('attribute_names')
        _IDX = [ATTR_NAMES.index(n) for n in names] if names and set(names) == set(ATTR_NAMES) else list(range(40))
    arrs = list(_POOL.map(_load, [s['image_path'] for s in samples]))
    x = torch.from_numpy(np.stack(arrs)).permute(0, 3, 1, 2).to(DEVICE)
    x = ((x - _MEAN) / _STD).contiguous(memory_format=torch.channels_last)
    with torch.autocast('cuda', dtype=torch.float16, enabled=(DEVICE.type == 'cuda')):
        lg = _MODEL(x).float() + _MODEL(torch.flip(x, dims=[3])).float()       # flip TTA
    p = torch.sigmoid(lg / 2 - _SHIFT).clamp(0, 1)
    p = torch.nan_to_num(p, nan=0.0)[:, _IDX]
    return p.cpu().numpy().astype(float).tolist()
```

```python
ref_model = PARNet(MODEL_NAME, pretrained=False).to(device).eval()
ref_model.load_state_dict(best['ema'])
n_test = 8
imgs = torch.stack([eval_tf(Image.open(p).convert('RGB')) for p in ho['paths'][:n_test]]).to(device)
with torch.no_grad():
    lg = (ref_model(imgs) + ref_model(torch.flip(imgs, dims=[3]))) / 2
    shift = torch.logit(torch.from_numpy(th).clamp(0.05, 0.95)).to(device)
    ref = torch.sigmoid(lg - shift).cpu().numpy()
np.save('/content/test_ref.npy', ref)
json.dump([str(p) for p in ho['paths'][:n_test]], open('/content/test_paths.json', 'w'))
json.dump(ATTR_NAMES, open('/content/attr_names.json', 'w'))
print('reference saved; predicted-positive rate on these images:', float((ref > 0.5).mean()))
```
